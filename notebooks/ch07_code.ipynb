## 7.2　GitHubとは：コードを置くクラウドのサービス

```bash
git clone https://github.com/kewel-corp/book-vol1.git    # 本書のコードを手元にコピー
cd book-vol1                       # clone は子フォルダを作るだけ。中へ入ってから作業する
# （ファイルを編集したあと）
git add .                          # 変更を記録の対象にする
git commit -m "OCT分類のコードを追加"   # 変更を履歴に記録する（-mはメッセージ）
git push                           # 自分が書き込める先へ送る（共有・バックアップ）
git pull                           # 他人の変更を取り込む
```

## 7.3　医療AIで必ず守ること：患者データはGitHubに絶対にアップロードしない

```text
# .gitignore の例（注釈は必ず別の行に書く。パターンの後ろに書くと、その注釈ごと
# 一つのパターンになり、除外が効かない）
# 画像データのフォルダは全体を除外
data/
# 推論結果・切り出し画像・図の出力先
outputs/
# DICOM画像・NIfTI画像
*.dcm
*.nii.gz
# 切り出した画像・オーバーレイ（患者由来の場合は漏えいに注意）
*.png
*.jpg
# ★対応表。患者ID・生年月日・撮影日を含むことがあるため、漏えいに注意
*.csv
# ★同上。Excelの対応表も、画像と同様に漏えいに注意
*.xlsx
# メタデータ（施設名・撮影条件が入ることがある）
*.json
# ★APIキー・トークン（Kaggleの認証情報や .env は、誤って公開される典型例）
kaggle.json
.env
*.key
# 学習済みモデルの重み（大きく、成果物でもある）
*.pth
```
